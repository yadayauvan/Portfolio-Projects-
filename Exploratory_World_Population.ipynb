```python
# Comparing column to every other column to show how correlated they are 
# Correlation means they are steadily increasing 

df.corr(numeric_only = True)
```

```python
# Heatmap to show the visualise the correlation table above 
# High correlation score are those tables two same tables correlated 

sns.heatmap(df.corr(numeric_only = True), annot=True)
plt.rcParams ['figure.figsize'] = (10,6)
plt.show()
```
